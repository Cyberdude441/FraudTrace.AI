## 1. Forensic Correlation Principles
- Evidence items are grouped by temporal proximity, monetary correlation, and entity linkage.
- **CRITICAL**: Distinct source observations are never merged or destroyed into a single synthetic record; each source observation retains its own provenance and audit trail.

```python
from ml.src.correlation.event_matching import EventMatcher
from ml.src.correlation.corroboration import CorroborationEngine

matcher = EventMatcher(time_window_minutes=60, amount_tolerance_ratio=0.10)
corrob_engine = CorroborationEngine()

sample_incident_evidence = [
    {
        "id": "EV-1",
        "type": "BANK_STATEMENT",
        "source": "HDFC Core",
        "amount": 50000.0,
        "timestamp": "2024-03-15T10:30:00Z",
        "entities": [{"type": "phone", "value": "9876543210"}]
    },
    {
        "id": "EV-2",
        "type": "SMS_GATEWAY",
        "source": "Telecom SMS",
        "amount": 50000.0,
        "timestamp": "2024-03-15T10:31:00Z",
        "entities": [{"type": "phone", "value": "9876543210"}]
    },
    {
        "id": "EV-3",
        "type": "POLICE_COMPLAINT",
        "source": "Victim FIR",
        "amount": 50000.0,
        "timestamp": "2024-03-15T10:30:00Z",
        "entities": [{"type": "phone", "value": "9876543210"}]
    },
    {
        "id": "EV-4",
        "type": "WHATSAPP_EXPORT",
        "source": "Chat Backup",
        "amount": 2500.0,
        "timestamp": "2024-03-15T16:45:00Z",
        "entities": [{"type": "upi", "value": "helper@upi"}]
    }
]

events = matcher.cluster_evidence(sample_incident_evidence)
print(f"Reconstructed {len(events)} Distinct Events from {len(sample_incident_evidence)} Evidence Items:\n")

for ev in events:
    sources = corrob_engine.get_source_count(ev['evidence_items'])
    indep = corrob_engine.get_independent_source_count(ev['evidence_items'])
    print(f"Event ID : {ev['event_id']}")
    print(f" - Timestamp Range      : {ev['start_time']} to {ev['end_time']}")
    print(f" - Mean Amount          : INR {ev['mean_amount']}")
    print(f" - Evidence Count       : {len(ev['evidence_items'])}")
    print(f" - Independent Sources  : {indep}")
    print(f" - Evidence IDs Included: {ev['evidence_ids']}\n")
```
