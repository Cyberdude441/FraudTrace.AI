## 1. Challenge Scenario: The Disputed Transfer
A complainant files an FIR alleging a fraudulent debit of **₹50,000** to phone number `+91 98765-43210` at 10:30 AM.
However, official HDFC core banking statements record a debit of **₹48,000** to the same phone entity `09876543210` at 10:32 AM.

### The Forensic Questions:
1. Are both records referring to the same phone entity?
2. Do the monetary figures corroborate or conflict?
3. What should the overall confidence level be?

```python
from ml.src.inference.pipeline import ForensicInferencePipeline
import json

pipeline = ForensicInferencePipeline()

evidence_items = [
    {
        "id": "EV-FIR-101",
        "type": "POLICE_COMPLAINT",
        "source": "Complainant Sworn Statement",
        "amount": 50000.0,
        "timestamp": "2024-03-15T10:30:00Z",
        "entities": [
            {"type": "phone", "value": "+91 98765-43210"}
        ],
        "text": "Complainant stated Rs. 50,000 was debited via UPI to recipient number 9876543210."
    },
    {
        "id": "EV-BANK-202",
        "type": "BANK_STATEMENT",
        "source": "HDFC Core Banking API Logs",
        "amount": 48000.0,
        "timestamp": "2024-03-15T10:32:00Z",
        "entities": [
            {"type": "phone", "value": "09876543210"}
        ],
        "text": "NEFT/IMPS outgoing transfer INR 48,000.00 to account linked with 9876543210."
    }
]

output = pipeline.analyze_event(evidence_items)
print("=== Forensic Inference Output JSON ===")
print(json.dumps(output, indent=2))
```

2. **Contradiction Engine**: Identified discrepancy between ₹50,000 and ₹48,000 (delta ₹2,000 / 4.0%). Classified as `CONFLICTING`.
3. **Confidence Scorer**: Guardrail triggered. Despite matching entities and timing, confidence is capped at `0.25 (LOW)`.
